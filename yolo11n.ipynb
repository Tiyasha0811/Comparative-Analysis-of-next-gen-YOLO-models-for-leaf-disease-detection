```python
import matplotlib.pyplot as plt

classes = ["Bougainvillea", "Bougainvillea Disease", "Ixora", "Ixora Disease"]
mAP50 = [0.995, 0.574, 0.995, 0.744]

plt.figure(figsize=(8, 5))
plt.bar(classes, mAP50, color='skyblue')
plt.title('mAP50 per Class')
plt.ylabel('mAP50')
plt.ylim(0, 1.1)
plt.grid(axis='y')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

results = pd.read_csv('/content/runs/segmentation/yolov8x-seg-10x/results.csv')  # path to your results
results[['epoch', 'train/box_loss', 'train/cls_loss']].plot(x='epoch', figsize=(10, 5))
plt.title('Training Loss per Epoch')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.grid(True)
plt.show()
```
